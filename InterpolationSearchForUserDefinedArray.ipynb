```python
import time
import random

def interpolation_search(arr, key):
    low = 0
    high = len(arr) - 1
    probes = 0

    while low <= high and arr[low] <= key <= arr[high]:

        probes += 1

        if low == high:
            if arr[low] == key:
                return low, probes
            return -1, probes

        if arr[high] == arr[low]:
            break

        pos = low + int(
            ((key - arr[low]) * (high - low))
            / (arr[high] - arr[low])
        )

        if arr[pos] == key:
            return pos, probes

        elif arr[pos] < key:
            low = pos + 1

        else:
            high = pos - 1

    return -1, probes


def manual_dataset():
    n = int(input("Enter number of elements: "))

    arr = []
    print("Enter elements:")

    for i in range(n):
        arr.append(int(input()))

    arr.sort()
    return arr


def random_uniform_dataset(n):
    return sorted(random.sample(range(1, n * 10), n))


def random_nonuniform_dataset(n):
    arr = []

    current = 1

    for i in range(n):
        arr.append(current)
        current += random.randint(1, 50)

    return arr


def perform_search(arr):
    print("\nDataset:")
    print(arr)

    key = int(input("\nEnter search key: "))

    start_time = time.perf_counter()

    index, probes = interpolation_search(arr, key)

    end_time = time.perf_counter()

    execution_time = (end_time - start_time) * 1000

    print("\n----- Search Result -----")

    if index != -1:
        print("Key Found")
        print("Index Position:", index)
    else:
        print("Key Not Found")

    print("Number of Probes/Comparisons:", probes)
    print(f"Execution Time: {execution_time:.6f} ms")

    return probes, execution_time


# ---------------- MAIN PROGRAM ----------------

print("INTERPOLATION SEARCH ON USER-DEFINED DATASETS")
print("\n1. Uniform Dataset")
print("2. Non-Uniform Dataset")

dataset_choice = int(input("\nChoose Dataset Type (1/2): "))

print("\n1. Manual Input")
print("2. Random Generation")

input_choice = int(input("\nChoose Data Entry Method (1/2): "))

if input_choice == 1:

    arr = manual_dataset()

else:
    n = int(input("Enter dataset size: "))

    if dataset_choice == 1:
        arr = random_uniform_dataset(n)
    else:
        arr = random_nonuniform_dataset(n)

probes, exec_time = perform_search(arr)

print("\n----- Performance Summary -----")

if dataset_choice == 1:
    print("Dataset Type: Uniform")
    print("Interpolation Search performs efficiently on uniformly distributed data.")
else:
    print("Dataset Type: Non-Uniform")
    print("Interpolation Search may require more probes on non-uniform data.")

print("Total Probes:", probes)
print(f"Execution Time: {exec_time:.6f} ms")
```
